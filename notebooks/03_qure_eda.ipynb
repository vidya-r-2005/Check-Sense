# 03. Exploratory Data Analysis: QuRE External Test Set

Analyzes the QuRE dataset (Femmer et al., 2025; Zenodo DOI 10.5281/zenodo.15656471).
Serves as the external test set for requirement ambiguity and smell detection.


```python
from pathlib import Path

import pandas as pd

qure_path = Path('../data/raw/qure/QuRE.csv')
df = pd.read_csv(qure_path)
print(f'Total requirements: {len(df):,}')
print(f'Columns: {list(df.columns)}')
df[['id', 'defect', 'weak_word']].head()
```

## 2. Defect vs OK Label Distribution


```python
defect_counts = df['defect'].value_counts()
pd.DataFrame({'Count': defect_counts, 'Percentage': (defect_counts / len(df) * 100).round(2)})
```

## 3. Weak Word Frequency (Top 15)


```python
weak_words = df['weak_word'].value_counts().head(15)
pd.DataFrame({'Count': weak_words, 'Percentage': (weak_words / len(df) * 100).round(2)})
```

## 4. 20 Random Samples (Seed 42) - Metadata Only


```python
sample_20 = df.sample(n=20, random_state=42)[['id', 'weak_word', 'defect']].reset_index(drop=True)
sample_20
```
